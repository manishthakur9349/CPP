```python
import pandas as pd
from matplotlib import pyplot as plt
```

```python
df = pd.read_csv("/content/TeachingRatings.csv")

print(df.head())
```

```python
# Q1 :- Create a percentage table for minority and tenure

minority_tenure = pd.crosstab(
    df["minority"],
    df["tenure"],
    normalize="index"
) * 100

print(minority_tenure)

visible_minority = df[df["minority"] == "yes"]

percentage = (visible_minority["tenure"] == "yes").mean() * 100

print("Percentage of visible minority professors who are tenured:",
      percentage)
```

```python
# Q2 :- Mean and standard deviation of age by tenure

age_stats = df.groupby("tenure")["age"].agg(["mean", "std"])

print(age_stats.round(2))
```

```python
# Q3 :- Graph for the age variable
plt.hist(df["age"], bins=10, edgecolor="black")

plt.xlabel("Age")
plt.ylabel("Number of Professors")
plt.title("Distribution of Professor Ages")

plt.show()
```

```python
# Q4 :-

gender_counts = df["gender"].value_counts()

print(gender_counts)

plt.bar(gender_counts.index, gender_counts.values)

plt.xlabel("Gender")
plt.ylabel("Number of Professors")
plt.title("Number of Professors by Gender")

plt.show()

plt.barh(gender_counts.index, gender_counts.values)

plt.xlabel("Number of Professors")
plt.ylabel("Gender")
plt.title("Number of Professors by Gender")

plt.show()
```

```python
# Q5 :- Median evaluation score for tenured professors
tenured = df[df["tenure"] == "yes"]

median_eval = tenured["eval"].median()

print("Median evaluation score for tenured professors:",
      median_eval)
```
